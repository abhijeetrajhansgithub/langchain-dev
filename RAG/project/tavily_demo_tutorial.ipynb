```python
import os
import asyncio
import ssl 
from dotenv import load_dotenv, find_dotenv
import certifi
import json 
from datetime import datetime

from langchain_tavily import TavilyExtract, TavilyMap
from rich.console import Console
from rich.panel import Panel
from rich.table import Table
from rich.progress import Progress

load_dotenv(find_dotenv())

# Configure SSL contexts
ssl_context = ssl.create_default_context(cafile=certifi.where())
os.environ['SSL_CERT_FILE'] = certifi.where()
os.environ['REQUESTS_CA_BUNDLE'] = certifi.where()

# Initialize console
console = Console()


print("All imports done!")
```

```python
tavily_map = TavilyMap(
    max_depth=3,
    max_breadth=15,
    limit=500
)

print("tavily_map initialized!")
```

```python
demo_url = "https://python.langchain.com/docs/introduction/"

site_map = tavily_map.invoke(demo_url)  # type: ignore
print("site_map: ", site_map)

urls = site_map.get("results", [])
console.print(f"Found {len(urls)} URLs", style="bold green")

# show first 10 as examples
for i, url in enumerate(urls[:10]):
    console.print(f"{i+1}. {url}", style="bold green")


if len(urls) > 10:
    console.print(f"   ... and {len(urls) - 10} more URLs", style="bold green")
```

```python
tavily_extract = TavilyExtract()

print("tavily_extract initialized!")
```

```python
sample_urls = urls[:20]

console.print(f"Extracting info from {len(sample_urls)} URLs", style="bold green")

extraction_result = await tavily_extract.ainvoke(input={'urls': sample_urls})  # type: ignore

extracted_docs = extraction_result.get("results", [])

console.print(f"Extracted {len(extracted_docs)} documents", style="bold green")

# Show summary of each extracted document
for i, doc in enumerate(extracted_docs, 1):
    url = doc.get("url", "Unknown URL")
    content = doc.get("raw_content", "No content found")

    panel_content = f"""
    URL: {url}
    Content Length: {len(content)}
    Preview: {content[:250]}
    """
    console.print(Panel.fit(panel_content, title=f"Document {i}", border_style="blue"))
```

### Batch Process

```python
from typing import List, Dict, Any
```

```python
def chunk_urls(urls: List[str], chunk_size: int = 3) -> List[List[str]]:
    chunks: List[List[str]] = []

    for i in range(0, len(urls), chunk_size):
        chunk = urls[i:i + chunk_size]
        chunks.append(chunk)
    
    return chunks


async def extract_batch(urls: List[str], batch_num: int) -> List[Dict[str, Any]]:
    try: 
        console.print(f"Processing batch {batch_num} with {len(urls)} URLs", style="bold green")
        docs = await tavily_extract.ainvoke(input={'urls': urls})  # type: ignore
        results = docs.get("results", [])
        console.print(f"Extracted {len(results)} documents from batch {batch_num}", style="bold green")
        return results
    
    except Exception as e:
        console.print(f"Error processing batch {batch_num}: {e}", style="bold red")
        return []
    

url_batches = chunk_urls(urls[:9], chunk_size=3)

console.print("Extracting 9 URLS...")
console.print("URL Batches: ", len(url_batches))

tasks = [extract_batch(batch, i+1) for i, batch in enumerate(url_batches)]
batch_results = await asyncio.gather(*tasks)

# Flatten the results
all_extracted: List[Dict[str, Any]] = []

for batch_result in batch_results:
    all_extracted.extend(batch_result)

console.print(f"Extracted {len(all_extracted)} documents from {len(urls[:9])} URLs", style="bold green")
```
